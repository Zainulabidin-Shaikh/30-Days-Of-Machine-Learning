## Text Cleaning

 1. Sample 10000 rows
 2. Remove html tags
 3. Remove special characters
 4. Converting every thing to lower case
 5. Removing Stop words
 6. Stemming

```python
# This cell samples 10000 rows randomly from the DataFrame for processing
df = df.sample(10000)
```

```python
df.shape
```

```python
df.info()
```

```python
# 'positive' is replaced with 1 and 'negative' with 0, inplace modifies the original DataFrame

df['sentiment'].replace({'positive': 1, 'negative': 0}, inplace=True)
```

```python
df.head()
```

```python
# a regular expression to remove HTML tags and tests it on the third review
import re
clean = re.compile('<.*?>')
re.sub(clean, '', df.iloc[2].review)
```

```python
# Defining a function to clean HTML tags from text
def clean_html(text):
    clean = re.compile('<.*?>')
    return re.sub(clean, '', text)
```

```python
# Applying clean_html function to the 'review' column of the DataFrame
df['review'] = df['review'].apply(clean_html)
```

```python
def convert_lower(text):
    return text.lower()
```

```python
# applying the convert_lower function to the 'review' column
df['review'] = df['review'].apply(convert_lower)
```

```python
# defining a function to remove special characters, replacing them with spaces
def remove_special(text):
    x = ''
    for i in text:
        if i.isalnum():
            x = x + i
        else:
            x = x + ' '
    return x
```

```python
# test the remove_special function with a sample text
remove_special(' th%e @ classic use of the word.it is called oz as that is the nickname given to the oswald maximum security state penitentary. it focuses mainly on emerald city, an experimental section of the prison where all the cells have glass fronts and face inwards, so privacy is not high on the agenda. em city is home to many..aryans, muslims, gangstas, latinos, christians, italians, irish and more....so scuffles, death stares, dodgy dealings and shady agreements are never far away.i would say the main appeal of the show is due to the fact that it goes where other shows wouldnt dare. forget pretty pictures painted for mainstream audiences, f')
```

```python
# applying the remove_special function to the 'review' column
df['review'] = df['review'].apply(remove_special)
```

```python
import nltk
```

```python
from nltk.corpus import stopwords
```

```python
# display the list of English stopwords
stopwords.words('english')
```

```python
df
```

```python
#defining a function to remove stopwords from text
def remove_stopwords(text):
    x = []
    for i in text.split():
        if i not in stopwords.words('english'):
            x.append(i)
    y = x[:]
    x.clear()
    return y
```

```python
# applying the remove_stopwords function to the 'review' column
df['review'] = df['review'].apply(remove_stopwords)
```

```python
df
```

```python
#  import PorterStemmer from NLTK for word stemming
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
# defining a function to perform stemming on a list of words
y = []
def stem_words(text):
    for i in text:
        y.append(ps.stem(i))
    z = y[:]
    y.clear()
    return z
```

```python
# testing the stem_words function with a sample list of words
stem_words(['I', 'loved', 'loving', 'it'])
```

```python
# applying the stem_words function to the 'review' column
df['review'] = df['review'].apply(stem_words)
```

```python
df
```

```python
# defining a function to join a list of words back into a string
def join_back(list_input):
    return " ".join(list_input)
```

```python
# applying the join_back function to the 'review' column
df['review'] = df['review'].apply(join_back)
```
